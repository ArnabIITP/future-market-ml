# Walk-forward model evaluation
Run expanding-window validation, then compare error by fold to see whether results change over time. The high-volatility event threshold is recalculated from each training window.

```python
import sys
sys.path.insert(0, '..')
from pathlib import Path
import pandas as pd
from src.preprocessing import prepare_prices
from src.features import make_dataset
from src.evaluation import evaluate_models, evaluate_high_volatility
files = sorted(Path('../data').glob('*.csv'))
if not files: print('Download data by running the experiment first.')
else:
    raw = pd.read_csv(files[0])
    price_col = 'Close' if 'Close' in raw else 'Adj Close'
    X, y, y_vol = make_dataset(prepare_prices(raw, 'Date', price_col))
    predictions, metrics = evaluate_models(X, y, y_vol)
    display(metrics.groupby('model')[['mae','rmse']].mean().sort_values('mae'))
    display(metrics.pivot(index='fold', columns='model', values='mae'))
    event_predictions, event_metrics = evaluate_high_volatility(X, y_vol)
    display(event_metrics)
```
