# Exploratory analysis
Inspect one local daily price series before modeling. The first run uses the experiment pipeline's cached SPY file if available.

```python
from pathlib import Path
import pandas as pd
files = sorted(Path('../data').glob('*.csv'))
if not files: print('No local data yet. Run the experiment downloader from the repository root.')
else:
    prices = pd.read_csv(files[0], parse_dates=['Date']).set_index('Date')
    prices.head()
```

```python
if files:
    close = prices['Close'] if 'Close' in prices else prices['Adj Close']
    returns = close.pct_change().dropna()
    display(returns.describe(percentiles=[.01,.05,.5,.95,.99]))
    display(returns.rolling(21).std().dropna().describe())
```
